```python
import pandas as pd

klienci = pd.read_csv("data/klienci.csv")
klienci
```

```python
import seaborn as sns

sns.scatterplot(x=klienci["roczny_dochod"], y=klienci["wskaznik_wydatkow"])
```

# Standaryzacja

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
x = klienci[["roczny_dochod", "wskaznik_wydatkow"]]
z = scaler.fit_transform(x)
```

# Metoda k-średnich

```python
from sklearn.cluster import KMeans

kmeans_klienci = KMeans(n_clusters=5, random_state=42)
grupy_klienci = kmeans_klienci.fit_predict(z)
grupy_klienci

klienci["grupa_kmeans"] = grupy_klienci

sns.scatterplot(x=klienci["roczny_dochod"], y=klienci["wskaznik_wydatkow"], hue=klienci["grupa_kmeans"], palette="Set2")

klienci["grupa_kmeans"].value_counts()
```

## Wybór liczby grup

```python
from sklearn.metrics import calinski_harabasz_score

inertia = []
ch_scores = []

for k in range(2, 11):
    kmeans = KMeans(n_clusters=k, random_state=42)
    labels = kmeans.fit_predict(z)
    inertia.append(kmeans.inertia_)
    ch_scores.append(calinski_harabasz_score(z, labels))

print("Inertia:", inertia)
print("Calinski-Harabasz Scores:", ch_scores)
```

```python
import matplotlib.pyplot as plt

plt.plot(range(2, 11), inertia, marker="o")
```

```python
plt.plot(range(2, 11), ch_scores, marker="o")
```

## Zadanie auta

```python
auta = pd.read_csv("data/auta.csv", sep=";")

scaler_auta = StandardScaler()
z_auta = scaler_auta.fit_transform(auta[["pojemnosc","przebieg","lata","cena"]])
```

```python
auta[["pojemnosc","przebieg","lata","cena"]].corr()
```

```python
auta[["pojemnosc","przebieg","lata","cena"]].std()/auta[["pojemnosc","przebieg","lata","cena"]].mean()
```

```python
inertia_auta = []
ch_scores_auta = []

for k in range(2, 11):
    kmeans = KMeans(n_clusters=k, random_state=42)
    labels = kmeans.fit_predict(z_auta)
    inertia_auta.append(kmeans.inertia_)
    ch_scores_auta.append(calinski_harabasz_score(z_auta, labels))

print("Inertia:", inertia_auta)
print("Calinski-Harabasz Scores:", ch_scores_auta)

plt.plot(range(2, 11), inertia_auta, marker="o")
```

```python
plt.plot(range(2, 11), ch_scores_auta, marker="o")
```

```python
auta["grupa_kmeans"] = KMeans(n_clusters=3, random_state=42).fit_predict(z_auta)
auta[["pojemnosc","przebieg","lata","cena","grupa_kmeans"]].groupby("grupa_kmeans").mean()
```

# Metoda hierarchiczna

```python
from sklearn.cluster import AgglomerativeClustering

hier_klienci = AgglomerativeClustering(n_clusters=5, linkage="ward")
klienci["grupa_hier"] = hier_klienci.fit_predict(z)

sns.scatterplot(x=klienci["roczny_dochod"], y=klienci["wskaznik_wydatkow"], hue=klienci["grupa_hier"], palette="Set2")
```

## Dendrogram

```python
from scipy.cluster.hierarchy import dendrogram, linkage

linkage_matrix = linkage(z, method="ward")

dendrogram(linkage_matrix, color_threshold=5)
```

## Zadanie - dendrogram

```python
auta_linked = linkage(z_auta, method="ward")

dendrogram(auta_linked, orientation="right", labels=auta["model"].values)
```

# Redukcja wymiarowości

```python
pca.explained_variance_ratio_
```

```python
sum(pca.explained_variance_ratio_)
```

```python
sns.scatterplot(x=z_auta_pca[:, 0], y=z_auta_pca[:, 1], hue=auta["grupa_kmeans"], palette="Set2")
```
